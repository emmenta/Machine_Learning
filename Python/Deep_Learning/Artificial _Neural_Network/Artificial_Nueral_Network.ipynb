# Artificial Neural Network

## Import the Libraries

```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
tf.__version__
```

## PART 1 - Data Preprocessing

### Import DataSet

```python
data = pd.read_csv('Churn_Modelling.csv')
print(data)
```

```python
x = data.iloc[:, 3:-1].values
y = data.iloc[:, -1].values
```

```python
print(x)
```

```python
print(y)
```

### Encoding Categorical Data

Label Encoding the "GENDER" column

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
x[:,2] = le.fit_transform(x[:,2])
```

```python
print(x)
```

One Hot Encoding the "GEOGRAPHY" column

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers =[('encoder', OneHotEncoder(), [1])], remainder = 'passthrough')
x = np.array(ct.fit_transform(x))
```

```python
print(x)
```

### Spliting the data into training and test set

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
```

### Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

## PART 2 - Buliding the ANN

### Initializing the ANN

```python
ann =  tf.keras.models.Sequential()
```

### Adding the input layer and the first hidden layer

```python
ann.add(tf.keras.layers.Dense(units = 6, activation = 'relu'))
```

### Adding the second hidden layer

```python
ann.add(tf.keras.layers.Dense(units = 6, activation = 'relu'))
```

### Adding the output layer

```python
ann.add(tf.keras.layers.Dense(units = 1, activation = 'sigmoid'))
```

## PART 3 - Training the ANN

### Compiling the ANN

```python
ann.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

### Training the ANN on the Training set

```python
ann.fit(x_train, y_train, batch_size = 32, epochs = 100)
```

## PART 4 - Making the predicitons and evaluating the model

### Predicting the result of a single observation

Use the ANN model to predict if the customer with the following informations will leave the bank:
- Geography: France
- Credit Score: 600
- Gender: Male
- Age: 40 Years
- Tenure: 3 Years
- Balance: \$ 60000
- Number of Products: 2
- Does this customer hae a credit card? Yes
- Is this customer an Active Member: Yes
- Estimated Salary: \$ 50000
  
So, should we say goodbye to that customer?

```python
print(ann.predict(sc.transform([[1, 0, 0, 600, 1, 40, 3, 60000, 2, 1, 1, 50000]])))
```

```python
print(ann.predict(sc.transform([[1, 0, 0, 600, 1, 40, 3, 60000, 2, 1, 1, 50000]])) > 0.5)
```

### Predicting the Test Result

```python
y_pred = ann.predict(x_test)
y_pred = (y_pred > 0.5)
print(np.concatenate((y_pred.reshape(len(y_pred), 1), y_test.reshape(len(y_test), 1)), 1))
```

### Making the Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```
